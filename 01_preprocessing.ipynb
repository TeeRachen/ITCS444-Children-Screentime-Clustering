#### Variables excluded before candidate selection

Some relevant variables were ruled out **before** the candidate list below, because they are not asked of every age group or are only asked conditionally. Including them would make whole age groups structurally missing, and that missingness cannot be meaningfully imputed. The codebook rules come from the *2024 NSCH Topical Variable List*; the missingness was checked in `children_topical.dta`.

| Variable | Codebook universe | Missing by form (T1 / T2 / T3) | Reason for exclusion |
|---|---|---|---|
| `height`, `weight` | T2 and T3 only | 100% / 5.3% / 2.1% (height), 100% / 2.6% / 1.7% (weight) | Not asked for ages 0–5 |
| `focuson` | T1 only, skipped if age < 3 | 46.9% / 100% / 100% | Asked only for ages 3–5 |
| `eyecare2` (eye doctor prescribed glasses or contact lenses) | All forms, skipped if `eyedoctor` = 2 (No) | 84.5% / 53.3% / 40.9% (56.4% overall) | Conditional on `eyedoctor`: answered only by children who saw an eye doctor, so it is structurally missing for everyone else |

Two further form-specific variables (`physactiv`, and the two sleep items `hoursleep` / `hoursleep05`) are part of the candidate list and are handled explicitly below.

```python
candidate_general = [
    "sc_age_years",   # Child's age (years)
    "sc_sex",         # Child's gender
    "family_r"        # Family structure
]

candidate_health = [
    "blindness",    # Blindness or serious vision difficulty
    "eyedoctor",    # Whether the child has seen an eye doctor
    "k4q22_r",      # Mental health professional treatment/need
    "k4q23"         # Medication due to emotional/concentration/behavior difficulties
]

candidate_lifestyle = [
    "screentime",
    "hoursleep",    # hour sleep for T2,T3
    "hoursleep05",  # hour sleep for T1
    "physactiv",    # Exercise, Play Sport, or Physical Activity for 60 Minutes
    "k6q71_r"       # Interest/curiosity in learning
]

candidate_environmet = [
    "higrade",      # Adult education
    "foodsit",      # Household food affordability
    "k10q12",       # Park/playground
    "k10q13",       # Recreation/community center
    "k10q14"        # Library/bookmobile
]
```

Checking the number of missing values

```python
def missing_summary(df, variables):
    return (
        pd.DataFrame({
            "Missing Count": df[variables].isna().sum(),
            "Missing Percentage": (
                df[variables].isna().mean() * 100
            ).round(2)
        })
        .sort_values("Missing Count", ascending=False)
    )


all_candidate_vars = (
    candidate_general
    + candidate_health
    + candidate_lifestyle
    + candidate_environmet
)

missing = missing_summary(df_topical, all_candidate_vars)

print(missing)
```

The **top 3 most missing values** are

2.   `physactiv`: 14349 missing (27.93%)
3.   `hoursleep`: 14336 missing (27.90%)

These high percentages are primarily due to variables being collected only for specific NSCH form types.

<br> Therefore, the missingness was examined by form type before deciding how to handle these variables.

<br> `hoursleep05` is collected for T1 respondents, while `hoursleep` is collected for T2 and T3 respondents. Therefore, these age-specific sleep variables were combined into `sleep_hours` variable.

#### Sleep: the two source variables use different hour scales

From the *2024 NSCH Topical Variable List* (U.S. Census Bureau):

| Code | `HOURSLEEP05` (T1, ages 0–5): "during an average day (count both nighttime sleep and naps)" | `HOURSLEEP` (T2/T3, ages 6–17): "on most weeknights" |
|---|---|---|
| 1 | Less than 7 hours | Less than 6 hours |
| 2 | 7 hours | 6 hours |
| 3 | 8 hours | 7 hours |
| 4 | 9 hours | 8 hours |
| 5 | 10 hours | 9 hours |
| 6 | 11 hours | 10 hours |
| 7 | 12 or more hours | 11 or more hours |

**Comparability review.** The two items differ in three ways:
1. **Hour anchors:** the same code means one hour more sleep on T1, and the open-ended categories differ.
2. **What is measured:** T1 asks about total sleep including naps on an average day; T2/T3 asks about sleep on most weeknights.
3. **Observed pattern:** within each form the mean code falls with age, but it jumps up at the form boundary (4.72 at age 5, 5.25 at age 6). Implied hours, by contrast, fall smoothly with age.

**The combined `sleep_hours` codes are therefore not comparable across forms.** They are kept in the export for traceability and descriptive EDA only, and are **not used as a clustering feature**.

**Was a below / within / above age-recommended recode possible?** The age-specific recommendations used by the NSCH Data Resource Center (indicator "child sleeps recommended age-appropriate hours", ages 4 months–17 years; AASM consensus) are: 4–12 months 12–16 h; 1–2 y 11–14 h; 3–5 y 10–13 h; 6–12 y 9–12 h; 13–17 y 8–10 h. Mapping the response codes onto these ranges gives:

| Age | Below | Within | Above |
|---|---|---|---|
| < 4 months | no recommendation exists | – | – |
| 4–11 months | codes 1–6 | code 7 (12+ h) is **ambiguous**: 12–16 h is within, 17+ h is above | |
| 1–2 y | codes 1–5 | code 6 | code 7 (12+ h) is **ambiguous**: 12–14 h is within, 15+ h is above |
| 3–5 y | codes 1–4 | codes 5–6 | code 7 (12+ h) is **ambiguous**: 12–13 h is within, 14+ h is above |
| 6–12 y | codes 1–4 | codes 5–6 | code 7 (11+ h) is **ambiguous**: 11–12 h is within, 13+ h is above |
| 13–17 y | codes 1–3 | codes 4–6 | code 7 |

The open-ended top category means the **"above" level cannot be identified reliably for ages 0–12**, so a three-level recode is not used.

**Primary sleep feature: `sleep_below_minimum`.** The "below" column is unambiguous for every child aged 4 months or older, because each age-specific minimum falls exactly on a code boundary. Each child is judged against the recommendation for their own age, which makes the indicator comparable across forms; the NSCH Data Resource Center pools the two items into one indicator in the same way. It is created in Step 3. Exact mapping (1 = below minimum, 0 = at or above):

| Age | Form | Minimum (h) | `sleep_hours` codes → 1 (below) | codes → 0 (at or above) |
|---|---|---|---|---|
| < 4 months (`sc_age_lt4` = 1) | T1 | none | missing (structural: no recommendation applies) | missing |
| 4–11 months | T1 | 12 | 1–6 (< 7 h … 11 h) | 7 (12+ h) |
| 1–2 y | T1 | 11 | 1–5 (< 7 h … 10 h) | 6–7 (11 h, 12+ h) |
| 3–5 y | T1 | 10 | 1–4 (< 7 h … 9 h) | 5–7 (10 h … 12+ h) |
| 6–12 y | T2/T3 | 9 | 1–4 (< 6 h … 8 h) | 5–7 (9 h … 11+ h) |
| 13–17 y | T3 | 8 | 1–3 (< 6 h … 7 h) | 4–7 (8 h … 11+ h) |

Caveats:
- **Overlap with age:** the indicator is defined relative to age, and `sc_age_years` is also a clustering feature, so some age-related information overlaps.
- **Guideline steps:** rates change abruptly where the guideline changes. For example, 50.8% are below the minimum at age 12 vs 14.2% at age 13, when the minimum drops from 9 h to 8 h. This is a property of the guideline, not of children's sleep.
- **Imputed values:** for the 627 children whose `sleep_hours` was imputed, the indicator is derived from the imputed value (tracked by `sleep_hours_imputed`).
- **Infants under 4 months:** their missing value is structural (no recommendation applies), not nonresponse, and is not imputed.

Because the scales differ, missing `sleep_hours` values are imputed **separately within the T1 scale and the T2/T3 scale** (see Step 2, imputation).

```python
def missing_by_group(df, variables, group_col):
    return (
        df.groupby(group_col)[variables]
        .apply(lambda group: group.isna().mean() * 100)
        .T
        .round(2)
    )


most_missing_vars = missing[missing["Missing Count"] > 1500].index.tolist()

print(missing_by_group(df_topical, most_missing_vars, "formtype"))
```

```python
df_topical["sleep_hours"] = df_topical["hoursleep"].fillna(
    df_topical["hoursleep05"]
)

missing_sleep_hours = missing_summary(
    df_topical,
    ["sleep_hours"]
)

print(missing_sleep_hours)
```

Most candidate variables are **categorical** or **ordinal**.

<br> Therefore, **mode imputation** is used to handle the remaining small amount of missing data.

The mode of each variable is calculated from its observed values and used to replace its missing values.

- **Exception, `sleep_hours`:** T1 and T2/T3 codes are on different hour scales (see above), so the mode is calculated **separately for T1 and for T2/T3** and each group is filled with its own mode.
- **Missingness indicators:** before imputing, an indicator column `<variable>_imputed` (1 = value was missing and imputed, 0 = observed) is created for every imputed variable. These indicators are exported for transparency and for a complete-case sensitivity check. They are **not clustering features**.

```python
impute_vars = [
    var
    for var in all_candidate_vars
    if df_topical[var].isna().any()
]

print(impute_vars)

# Missingness indicators, recorded BEFORE imputation (1 = imputed, 0 = observed).
# Kept in a separate DataFrame so they never enter the clustering feature list.
missing_flags = df_topical[impute_vars].isna().astype("int8")
missing_flags.columns = [f"{var}_imputed" for var in impute_vars]
missing_flag_cols = missing_flags.columns.tolist()

print("\nImputed values per variable:")
print(missing_flags.sum())
print("\nChildren with at least one imputed value:", int(missing_flags.any(axis=1).sum()))
print("Complete cases (no imputed values):", int((~missing_flags.any(axis=1)).sum()))
```

```python
for var in impute_vars:
    if var == "sleep_hours":
        # T1 (hoursleep05) and T2/T3 (hoursleep) codes are on different hour scales,
        # so impute each scale group with its own mode.
        sleep_groups = {
            "T1": df_topical["formtype"] == "T1",
            "T2/T3": df_topical["formtype"].isin(["T2", "T3"])
        }
        for group_name, mask in sleep_groups.items():
            mode_value = df_topical.loc[mask, var].mode()[0]
            n_missing = int(df_topical.loc[mask, var].isna().sum())
            df_topical.loc[mask, var] = df_topical.loc[mask, var].fillna(mode_value)
            print(f"sleep_hours ({group_name}): {n_missing} imputed with mode {mode_value}")
    else:
        mode_value = df_topical[var].mode()[0]
        df_topical[var] = df_topical[var].fillna(mode_value)

print(df_topical[all_candidate_vars].isna().sum())
```

### Step 3: Data Validation
Create a separate dataframe containing the selected candidate variables and verify that no missing values remain.

```python
# Processed candidate variables: the 15 variables selected and cleaned in this notebook (used by the EDA).
# NOTE: the FINAL clustering feature specification (14 features) is defined in 03_clustering.ipynb.
#       It uses sleep_below_minimum instead of sleep_hours and excludes eyedoctor (see the sleep notes above).
processed_features = all_candidate_vars.copy()

# Identifier and survey-design metadata (NOT clustering features).
#   hhid       - Unique Household ID; one sampled child per household, so it uniquely identifies each child
#   formtype   - Questionnaire form (T1 = ages 0-5, T2 = 6-11, T3 = 12-17)
#   fwc        - Selected Child Weight, for population-level (weighted) estimates
#   fipsst, stratum - State and sampling stratum; with hhid as PSU, needed for design-based standard errors
#   sc_age_lt4 - 1 = child is under 4 months old; identifies structural missingness in sleep_below_minimum
id_col = "hhid"
metadata_cols = ["formtype", "fwc", "fipsst", "stratum", "sc_age_lt4"]

# Derived sleep variable: the PRIMARY sleep feature for clustering (see 03_clustering.ipynb).
# sleep_below_minimum: 1 = sleep below the minimum recommended hours for the child's age,
#                      0 = at or above the minimum, missing = infant < 4 months (no recommendation applies).
# Minimum hours (AASM consensus, as used by the NSCH Data Resource Center):
#   4-11 months 12 h, 1-2 y 11 h, 3-5 y 10 h, 6-12 y 9 h, 13-17 y 8 h.
# Code-to-hours scales: T1 code k = k + 5 hours (code 1 = "<7 h"); T2/T3 code k = k + 4 hours (code 1 = "<6 h").
# A child is below the minimum if code < (minimum hours - scale offset). Every minimum falls on an exact
# code boundary, so this mapping is unambiguous (unlike a "within/above" split, see the sleep notes above).
derived_cols = ["sleep_below_minimum"]

age = df_topical["sc_age_years"]
min_hours = pd.cut(age, bins=[-1, 0, 2, 5, 12, 17], labels=[12, 11, 10, 9, 8]).astype(int)
scale_offset = df_topical["formtype"].map({"T1": 5, "T2": 4, "T3": 4})
below_min = (df_topical["sleep_hours"] < (min_hours - scale_offset)).astype("Int8")
under_4_months = (age == 0) & (df_topical["sc_age_lt4"] == 1)
below_min[under_4_months] = pd.NA
derived = pd.DataFrame({"sleep_below_minimum": below_min})

print("sleep_below_minimum:", derived["sleep_below_minimum"].value_counts(dropna=False).to_dict())
print(pd.crosstab([df_topical["formtype"], min_hours.rename("min_hours")],
                  [df_topical["sleep_hours"], derived["sleep_below_minimum"].fillna(-1)]))

df_cluster = df_topical[processed_features].copy()

assert df_topical[id_col].is_unique and df_topical[id_col].notna().all()
print(df_cluster.isna().sum())
```

### Step 4: Export preprocessed data
Save the cleaned dataset as a CSV file so it can be imported directly for EDA and clustering.

Column groups in `datasets/preprocessed_data.csv`:

| Group | Columns | Use |
|---|---|---|
| Identifier | `hhid` | Stable child identifier for joining results. **Not a feature.** |
| Processed candidate variables | the 15 variables in `processed_features` | Used by the EDA. The clustering model uses 14 features selected from these plus `sleep_below_minimum` (defined in `03_clustering.ipynb`). |
| Survey metadata | `formtype`, `fwc`, `fipsst`, `stratum`, `sc_age_lt4` | Form type, survey-design variables for weighted profiling, and the under-4-months flag. **Not features.** |
| Derived | `sleep_below_minimum` | 1 = below the age-specific minimum recommended sleep, 0 = at or above, blank = infant under 4 months (structural). **Primary sleep feature for clustering**, replacing the non-comparable `sleep_hours` codes. |
| Missingness indicators | `<variable>_imputed` | 1 = value was imputed. For transparency and complete-case sensitivity checks. **Not features.** |

Downstream notebooks should select features by name and never use all columns.

```python
# Assemble export: identifier, processed variables, survey metadata, derived variable, missingness indicators
df_export = pd.concat(
    [
        df_topical[[id_col]],
        df_cluster,
        df_topical[metadata_cols],
        derived,
        missing_flags
    ],
    axis=1
)

assert df_export[id_col].is_unique
assert df_export[processed_features].notna().all().all()
assert len(df_export) == len(df_topical)

# Save the preprocessed dataset for use in the EDA and clustering notebooks
output_path = Path.cwd() / "datasets" / "preprocessed_data.csv"
df_export.to_csv(output_path, index=False)
print("Saved:", output_path, df_export.shape)
print("Processed candidate variables:", len(processed_features), processed_features)
print("Metadata:", [id_col] + metadata_cols)
print("Derived:", derived_cols)
print("Missingness indicators:", missing_flag_cols)
```
